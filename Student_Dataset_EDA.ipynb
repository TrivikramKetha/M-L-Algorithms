```python
# Check for invalid academic values
print("Marks above 100:", (df["Marks"] > 100).sum())
print("Attendance above 100:", (df["Attendance"] > 100).sum())
print("Study Hours below 0:", (df["StudyHours"] < 0).sum())
```

```python
# Replace invalid values with NaN
df.loc[df["Marks"] > 100, "Marks"] = np.nan
df.loc[df["Attendance"] > 100, "Attendance"] = np.nan
df.loc[df["StudyHours"] < 0, "StudyHours"] = np.nan
```

```python
# Fill missing numerical values with their mean
df["StudyHours"] = df["StudyHours"].fillna(df["StudyHours"].mean())
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

df.isnull().sum()
```
