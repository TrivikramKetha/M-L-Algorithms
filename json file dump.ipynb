```python
import json
```

```python
my_dict = {
    "name": "Rahul",
    "age": 20,
    "course": "CSE",
    "college": "KL University"
}
```

```python
print(my_dict)
print(type(my_dict))
```

```python
y = json.dumps(my_dict)
print(y)
print(type(y))
```

```python
x = json.loads(y)
print(x)
print(type(x))
```

```python
y = json.dumps(my_dict, indent=4)
print(y)
```

```python
with open("student.json", "w") as f:
    json.dump(my_dict, f, indent=4)
```

```python
with open("student.json", "r") as f:
    data = json.load(f)

print(data)
print(type(data))
```
